#### What and Why Predictive Maintenance - Stack from broadest to most specific

1. **Artificial Intelligence** - Everything that makes machines simulate intelligent behaviour.
2. **Machine Learning** - The subset of AI where machines learn from data instead of being explicitly programmed.
3. **Applied ML / Industrial AI** - ML applied to real physical systems — factories, engines, aircraft, power grids. This is where your career lives.
4. **Condition-Based Monitoring (CBM)** - The engineering discipline of watching machines continuously and making decisions based on their actual condition, not a fixed schedule. Predates ML — engineers were doing this with statistical control charts before algorithms existed.
5. **Predictive Maintenance (PdM)** - A specific strategy within CBM. Instead of "maintain every 3 months" or "maintain when broken," you maintain when the data says it's about to break. ML is what made this actually work at scale.

### Learnings APR12261646

#### Time series data

- Each row is the same engine, one cycle later. Row 50 is the direct continuation of row 49. If you shuffle the rows, you destroy the entire meaning of the dataset. The past is everything. The trend is the signal.

#### Normal data

- Each row is a separate machine at a separate moment. Row 500 has nothing to do with row 499. You could shuffle all 10,000 rows randomly and your model would still train perfectly. The past doesn't exist. Only right now exists.

# Problem Statement 1 — Kaggle dataset [APR12262130]

Given a snapshot of sensor readings from an industrial machine at a single point in time — including temperature, rotational speed, torque, and tool wear — predict whether the machine will fail, and if so, classify the type of failure from five known failure modes: heat dissipation, power failure, overstrain, tool wear, or random failure.

1. **Type**: Multiclass classification
2. **Target**: Failure Type
3. **Challenge**: Severe class imbalance — 96.5% of rows are No Failure

# Problem Statement 2 — NASA turbofan dataset [APR12262130]

Given multivariate time series sensor readings recorded across the operational life of a turbofan aircraft engine — 21 sensors across multiple flight cycles — predict the Remaining Useful Life (RUL): the number of cycles remaining before the engine reaches failure.

1. **Type**: Regression
2. **Target**: RUL — to be engineered, not present in raw data
3. **Challenge**: No column headers, operating condition noise across 3 settings, 100 engines each with different lifespans
